```python
# Cell 1: Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src to path for imports
sys.path.append(os.path.abspath(os.path.join('..')))
from src import config, data_loader, features
```

```python
# Cell 2: Load Data
df = data_loader.download_data(config.TICKER, config.START_DATE, config.END_DATE)
df.head()
```

```python
# Cell 3: Feature Engineering
df = features.add_technical_indicators(df)
df.dropna(inplace=True)
```

```python
# Cell 4: Visualization - Price & Moving Averages
plt.figure(figsize=(14, 7))
plt.plot(df['Close'], label='Close Price', alpha=0.5)
plt.plot(df['SMA_50'], label='SMA 50', alpha=0.7)
plt.plot(df['SMA_200'], label='SMA 200', alpha=0.7)
plt.title(f"{config.TICKER} Price History with Moving Averages")
plt.legend()
plt.show()
```

```python
# Cell 5: Correlation Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=False, cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()
```

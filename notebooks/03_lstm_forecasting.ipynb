```python
# Cell 1: Imports
import numpy as np
import matplotlib.pyplot as plt
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..')))
from src import config, data_loader, lstm_model
```

```python
# Cell 2: Data Prep
df = data_loader.download_data(config.TICKER, config.START_DATE, config.END_DATE)
lstm = lstm_model.LSTMForecaster(sequence_length=60)
X, y, scaled = lstm.preprocess(df[['Close']])
```

```python
# Cell 3: Train/Test Split
split = int(len(X) * 0.8)
X_train, y_train = X[:split], y[:split]
X_test, y_test = X[split:], y[split:]
```

```python
# Cell 4: Build and Train
model = lstm.build_model(input_shape=(X_train.shape[1], 1))
history = lstm.train(X_train, y_train, validation_data=(X_test, y_test))
```

```python
# Cell 5: Loss Curves
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.legend()
plt.show()
```

```python
# Cell 6: Predictions
preds = lstm.predict(X_test)
actual = lstm.scaler.inverse_transform(y_test.reshape(-1, 1))

plt.figure(figsize=(12, 6))
plt.plot(actual, label='Actual')
plt.plot(preds, label='Predicted')
plt.title("LSTM Predictions")
plt.legend()
plt.show()
```
